## Задача 5. Исправить вложенную копию — повторение

Получить independent как глубокую копию profiles и добавить оценку 5 только первому ученику копии. Найти причину, по которой обычная copy здесь недостаточна.


```python
profiles = [{"name": "Ира", "marks": [4]}, {"name": "Лев", "marks": []}]
# РЕШЕНИЕ
independent = profiles.copy()
independent[0]["marks"].append(5)
```

```python
# Проверка
assert profiles == [{"name": "Ира", "marks": [4]}, {"name": "Лев", "marks": []}], "Исходные оценки изменились"
assert independent[0]["marks"] == [4, 5], "Копия должна содержать новую оценку"
assert independent[1]["marks"] == [], "Второго ученика не менять"
print("OK — задача 5")
```
