# Контрольная: коллекции и циклы

**Результат:** самостоятельно сочетать изученные структуры, условия и циклы на нескольких наборах данных.


## 1. Выбор структуры

Коллекция выбирается по задаче. list сохраняет порядок и повторы; tuple представляет фиксированные позиции; set отвечает за уникальность; dict хранит соответствия. Одна задача может сочетать их, например словарь счётчиков и список результатов.

Перед кодом нужно назвать вход, выход и правила: сохранять ли порядок, учитывать ли повторы, можно ли менять исходник. Без этих условий правильность невозможно проверить однозначно.


```python
events = ["save", "open", "save"]
counts = {}
for event in events:
    counts[event] = counts.get(event, 0) + 1
print(counts, sorted(counts))
```

**Практика:** Выбрать структуру для списка посещений и для набора уникальных посетителей.


## 2. Проверка алгоритма на границах

Пустой вход, повтор, ноль, отрицательное значение и равные кандидаты могут потребовать разных решений. Эти случаи не являются случайными препятствиями: они раскрывают правила задачи.

На этом этапе один и тот же алгоритм можно применить к нескольким наборам во внешнем цикле. Каждый набор должен получать заново инициализированные накопители. Позднее та же независимость будет оформляться функциями.


```python
cases = [[1, 2], [], [-1, 3]]
results = []
for values in cases:
    total = 0
    for value in values:
        total += value
    results.append(total)
print(results)
assert results == [3, 0, 2]
```

**Практика:** Показать, что сломается, если total задаётся только перед внешним циклом.


## 3. Порядок и независимость

Преобразование не должно незаметно менять вход, если это не разрешено. set убирает повторы, но не сохраняет требуемые позиции. Поверхностная копия не изолирует вложенные списки. Эти свойства нужно учитывать до выбора короткой записи.

Для отладки полезно сначала получить понятный правильный результат на небольшом наборе, затем проверить границы и упростить код. Сокращение количества строк не является самостоятельным критерием качества.


```python
values = ["b", "a", "b", "c"]
seen = set()
ordered_unique = []
for value in values:
    if value not in seen:
        seen.add(value)
        ordered_unique.append(value)
print(ordered_unique)
assert ordered_unique == ["b", "a", "c"]
```

**Практика:** Объяснить, зачем здесь одновременно список и множество.
