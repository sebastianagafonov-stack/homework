# Функции и return

**Результат:** выделять повторяемое действие в функцию и возвращать результат для дальнейшей обработки.


## 1. Определение и вызов

Функция — именованный блок кода, который вызывается для выполнения действия. Она нужна, чтобы описать правило один раз, применять к разным данным и проверять отдельно от остальной программы.

Определение начинается с def, затем идут имя, параметры в скобках и двоеточие. Тело имеет отступ. Определение само по себе не выполняет тело: оно создаёт функцию. Вызов `name(...)` выполняет тело с переданными значениями. Параметр — имя в определении, аргумент — переданное при вызове значение.


```python
def double(number):
    return number * 2

print(double(3))
print(double(8))
assert double(0) == 0
```

**Практика:** Показать отдельно строку определения и строки вызова.


## 2. return и print

return завершает текущий вызов и передаёт значение вызывающему коду. print только выводит текст и сам возвращает None. Если функция дошла до конца без return, её результат также None.

Возвращаемое число можно сложить с другим, строку — сохранить, список — обработать. Если результат только напечатан, такое повторное использование теряется. Поэтому вычисление обычно возвращает данные, а печать располагается снаружи, где действительно требуется показ.


```python
def area(width, height):
    result = width * height
    return result

first = area(3, 4)
combined = first + area(2, 5)
print(combined)
assert combined == 22
```

**Практика:** Перенести print внутрь функции вместо return и объяснить, почему сложение перестаёт работать; ошибочный вариант разбирать отдельно.


## 3. Ранний выход и несколько значений

Функция может завершиться раньше через return в условии или цикле. Это удобно, когда результат найден или вход требует специального ответа. Код после выполненного return в этом вызове не запускается.

Запись `return a, b` возвращает один объект — кортеж двух значений. Его можно распаковать снаружи. Все ветви функции должны согласованно возвращать результат, описанный в условии: случайное None на одной ветви обычно является ошибкой.


```python
def first_positive(values):
    for value in values:
        if value > 0:
            return value
    return None

def bounds(values):
    if not values:
        return None, None
    return min(values), max(values)

print(first_positive([-1, 0, 3, 5]))
print(bounds([]), bounds([3, 1, 8]))
```

**Практика:** Найти, чем return внутри поиска отличается от break.


## 4. Контракт и независимые вызовы

Контракт функции описывает вход, результат и важные ограничения. Он задаёт, можно ли менять вход и какой ответ нужен для пустого набора. Понятный контракт превращается в набор assert на нескольких входах.

Локальные накопители создаются заново при каждом вызове. Функция не должна случайно зависеть от значения, оставленного другой ячейкой. Короткая строка документации в тройных кавычках после def объясняет назначение; имена параметров должны отражать данные.


```python
def positive_sum(values):
    """Вернуть сумму положительных чисел без изменения входа."""
    total = 0
    for value in values:
        if value > 0:
            total += value
    return total

assert positive_sum([2, -1, 3]) == 5
assert positive_sum([]) == 0
assert positive_sum([-4]) == 0
```

**Практика:** Добавить к проверкам ноль и повторяющиеся значения.
