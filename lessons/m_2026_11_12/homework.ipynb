## Задача 1. Уникальные увлечения

Функция `unique_hobbies(hobbies)` получает список строк и возвращает **множество** увлечений без повторов.

Пример: `unique_hobbies(["шахматы", "плавание", "шахматы"])` → `{"шахматы", "плавание"}`.

Для пустого списка верни пустое множество. Порядок элементов множества не важен.

**Коротко:** `set(список)` создаёт множество без повторов. Пустое множество — `set()`, а `{}` — пустой словарь.
Материал занятия: [нужный раздел в lesson.ipynb](lesson.ipynb#sets).

```python
# ТВОЙ КОД
def unique_hobbies(hobbies):

    return set(hobbies)
```

```python
# Проверка — запускай после решения
assert isinstance(unique_hobbies(["шахматы"]), set), "Верни множество, а не список"
assert unique_hobbies(["шахматы", "плавание", "шахматы"]) == {"шахматы", "плавание"}, "Убери повторы"
assert unique_hobbies([]) == set(), "Для пустого списка нужно пустое множество"
assert unique_hobbies(["футбол", "футбол", "футбол"]) == {"футбол"}, "Одинаковое увлечение должно остаться один раз"
assert unique_hobbies(["бег", "рисование"]) == {"бег", "рисование"}, "Сохрани все разные увлечения"
original = ["бег", "бег", "шахматы"]
unique_hobbies(original)
assert original == ["бег", "бег", "шахматы"], "Не меняй исходный список"
print("Задача 1: верно")
```

## Задача 2. Общие игры

Функция `common_games(first, second)` получает два списка названий игр. Верни **множество** игр, которые есть в обоих списках.

Пример: `common_games(["Minecraft", "Portal"], ["Portal", "Terraria"])` → `{"Portal"}`.

Повторы в списках не должны влиять на ответ. Если общих игр нет, верни пустое множество.

**Коротко:** Преобразуй списки в множества. Пересечение `a & b` содержит только общие элементы.
Материал занятия: [нужный раздел в lesson.ipynb](lesson.ipynb#sets).

```python
# ТВОЙ КОД
def common_games(first, second):
    return set(first) & set(second)


```

```python
# Проверка — запускай после решения
assert isinstance(common_games(["Portal"], ["Portal"]), set), "Верни множество"
assert common_games(["Minecraft", "Portal"], ["Portal", "Terraria"]) == {"Portal"}, "Найди общую игру"
assert common_games(["Portal", "Portal", "Minecraft"], ["Portal", "Minecraft", "Minecraft"]) == {"Portal", "Minecraft"}, "Найди все общие игры без повторов"
assert common_games(["Portal"], ["Terraria"]) == set(), "Общих игр нет"
assert common_games([], ["Portal"]) == set(), "Первый список пуст"
assert common_games(["Portal"], []) == set(), "Второй список пуст"
assert common_games([], []) == set(), "Оба списка пусты"
first, second = ["Portal", "Portal"], ["Portal", "Terraria"]
common_games(first, second)
assert first == ["Portal", "Portal"] and second == ["Portal", "Terraria"], "Не меняй исходные списки"
print("Задача 2: верно")
```

## Задача 3. Телефонная книга

Функция `find_phone(phones, name)` получает словарь «имя → номер телефона» и имя для поиска. Номера хранятся строками.

Если имя есть в словаре, верни его номер. Иначе верни строку `"Нет номера"`.

Примеры:
- `find_phone({"Аня": "+7-111", "Боря": "+7-222"}, "Аня")` → `"+7-111"`;
- `find_phone({"Аня": "+7-111"}, "Вера")` → `"Нет номера"`.

Имена сравниваются точно: `"Аня"` и `"аня"` — разные ключи.

**Коротко:** `словарь.get(ключ, значение_по_умолчанию)` возвращает значение по ключу, а при отсутствии ключа — указанный запасной ответ.
Материал занятия: [нужный раздел в lesson.ipynb](lesson.ipynb#dicts).

```python
# ТВОЙ КОД
def find_phone(phones, name):

    return phones.get(name, "Нет номера")
```

```python
# Проверка — запускай после решения
assert find_phone({"Аня": "+7-111", "Боря": "+7-222"}, "Аня") == "+7-111", "Верни номер Ани"
assert find_phone({"Аня": "+7-111", "Боря": "+7-222"}, "Боря") == "+7-222", "Верни номер найденного человека"
assert find_phone({"Аня": "+7-111"}, "Вера") == "Нет номера", "Обработай отсутствующее имя"
assert find_phone({}, "Аня") == "Нет номера", "Телефонная книга может быть пустой"
assert find_phone({"Аня": "+7-111"}, "аня") == "Нет номера", "Не меняй регистр имени"
assert find_phone({"Боря": "00123"}, "Боря") == "00123", "Номер должен остаться строкой с ведущими нулями"
phones = {"Аня": "+7-111"}
find_phone(phones, "Вера")
assert phones == {"Аня": "+7-111"}, "Не добавляй отсутствующее имя в исходный словарь"
print("Задача 3: верно")
```

## Задача 4. Бонусные баллы

Функция `add_bonus(scores, bonus)` получает словарь «имя → баллы» и целое неотрицательное число бонусных баллов.

Верни **новый словарь**, в котором каждому участнику прибавлен `bonus`. Исходный словарь должен остаться прежним.

Пример: `add_bonus({"Аня": 10, "Боря": 7}, 3)` → `{"Аня": 13, "Боря": 10}`.

Для пустого словаря верни пустой словарь. При нулевом бонусе значения не изменяются.

**Коротко:** Создай результат через `{}`. Цикл `for name, points in scores.items()` перебирает пары «ключ, значение». Запись `result[name] = ...` добавляет значение в результат.
Материал занятия: [нужный раздел в lesson.ipynb](lesson.ipynb#dicts).

```python
# ТВОЙ КОД
def add_bonus(scores, bonus):
    result = {}
    for name, points in scores.items():
        result[name] = points + bonus
    return result
```

```python
# Проверка — запускай после решения
assert isinstance(add_bonus({"Аня": 10}, 3), dict), "Верни словарь"
assert add_bonus({"Аня": 10, "Боря": 7}, 3) == {"Аня": 13, "Боря": 10}, "Прибавь бонус каждому участнику"
assert add_bonus({"Вера": 0}, 5) == {"Вера": 5}, "Начальные баллы могут быть нулевыми"
assert add_bonus({}, 3) == {}, "Для пустого словаря верни пустой словарь"
scores = {"Аня": 10, "Боря": 7}
result = add_bonus(scores, 3)
assert scores == {"Аня": 10, "Боря": 7}, "Не меняй исходный словарь"
assert result is not scores, "Результат должен быть новым словарём"
zero_result = add_bonus(scores, 0)
assert zero_result == scores, "При нулевом бонусе баллы остаются прежними"
assert zero_result is not scores, "Даже при нулевом бонусе нужен новый словарь"
print("Задача 4: верно")
```

## Задача 5. Подсчёт фруктов

Функция `count_fruits(fruits)` получает список названий фруктов и возвращает словарь «фрукт → сколько раз он встретился».

Пример: `count_fruits(["яблоко", "банан", "яблоко"])` → `{"яблоко": 2, "банан": 1}`.

Для пустого списка верни пустой словарь. Названия сравниваются точно, без изменения регистра. Порядок ключей в ответе не важен.

**Коротко:** Перебирай список циклом. Для каждого фрукта увеличивай счётчик в словаре на 1. Если ключа ещё нет, его начальный счётчик равен 0; получить его можно через `.get(ключ, 0)`.
Материал занятия: [нужный раздел в lesson.ipynb](lesson.ipynb#dicts).

```python
# ТВОЙ КОД
def count_fruits(fruits):
    result = {}
    for fruit in fruits:
        result[fruit] = result.get(fruit, 0) + 1
    return result
```

```python
# Проверка — запускай после решения
assert isinstance(count_fruits(["яблоко"]), dict), "Верни словарь"
assert count_fruits(["яблоко", "банан", "яблоко"]) == {"яблоко": 2, "банан": 1}, "Посчитай каждый фрукт"
assert count_fruits([]) == {}, "Для пустого списка нужен пустой словарь"
assert count_fruits(["груша", "груша", "груша"]) == {"груша": 3}, "Посчитай все повторы"
assert count_fruits(["киви", "банан", "груша"]) == {"киви": 1, "банан": 1, "груша": 1}, "Каждый фрукт встретился один раз"
assert count_fruits(["Яблоко", "яблоко"]) == {"Яблоко": 1, "яблоко": 1}, "Не меняй регистр названий"
fruits = ["банан", "яблоко", "банан"]
count_fruits(fruits)
assert fruits == ["банан", "яблоко", "банан"], "Не меняй исходный список"
print("Задача 5: верно")
```
