# Часть 1. Теория

## 1. Список и индексы

Список — это несколько значений в одной переменной, по порядку, в квадратных скобках.
У каждого элемента есть номер — **индекс**. Нумерация начинается с **нуля**.

```
элемент:          5    4    5    3
индекс:           0    1    2    3
индекс с конца:  -4   -3   -2   -1
```

````python
marks = [5, 4, 5, 3]

print(marks[0])      # 5 — первый элемент
print(marks[-1])     # 3 — последний
print(len(marks))    # 4 — сколько элементов
print(sum(marks))    # 17 — сумма
print(max(marks))    # 5 — самый большой
print(sorted(marks)) # [3, 4, 5, 5] — отсортированная копия
````

## 2. Срезы — взять кусок списка

Срез: `список[начало:конец]`. **Начало входит, конец — НЕ входит.**

```
индекс:   0     1     2     3     4
список: ["пн", "вт", "ср", "чт", "пт"]
         [0:3] → берём индексы 0, 1, 2 → ["пн", "вт", "ср"]
```

````python
days = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]

print(days[0:3])   # ['пн', 'вт', 'ср']
print(days[2:5])   # ['ср', 'чт', 'пт']
````

Если не писать начало — берём с самого начала. Если не писать конец — берём до самого конца.

**Запомни две заготовки:**
- `список[:n]` — первые `n` элементов;
- `список[-n:]` — последние `n` элементов.

````python
days = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]

print(days[:2])    # ['пн', 'вт'] — первые 2
print(days[-2:])   # ['сб', 'вс'] — последние 2
````

## 3. Как изменить список

Список можно менять: добавлять, вставлять, удалять.

| Что сделать | Как |
|---|---|
| добавить в конец | `pets.append("хомяк")` |
| вставить на место с индексом `i` | `pets.insert(i, "рыбка")` |
| удалить по значению | `pets.remove("кот")` |
| заменить элемент | `pets[0] = "попугай"` |

````python
pets = ["кот", "собака"]

pets.append("хомяк")
print(pets)                # ['кот', 'собака', 'хомяк']

pets.insert(0, "рыбка")    # вставляем в начало, остальные сдвигаются вправо
print(pets)                # ['рыбка', 'кот', 'собака', 'хомяк']

pets.remove("кот")
print(pets)                # ['рыбка', 'собака', 'хомяк']

pets[0] = "попугай"
print(pets)                # ['попугай', 'собака', 'хомяк']
````

**Важно:** `append`, `insert`, `remove` меняют список сами и ничего не возвращают.
Писать `pets = pets.append("хомяк")` **нельзя** — в `pets` окажется `None`.
Правильно: просто `pets.append("хомяк")`, без `pets =`.

## 4. Есть ли элемент в списке: `in`

`"банан" in fruits` отвечает `True` или `False`.

````python
fruits = ["яблоко", "банан", "груша"]

print("банан" in fruits)       # True
print("киви" in fruits)        # False
print("киви" not in fruits)    # True
````

## 5. Условие `if` — «если»

Идея: **«если условие верно — сделай это»**.

```python
if условие:
    что делать, если условие верно
```

Три обязательные вещи:
1. слово `if`;
2. **двоеточие** `:` в конце строки;
3. **отступ** (4 пробела или Tab) у кода внутри. Всё, что с отступом, — внутри `if`.

````python
age = 15

if age >= 14:
    print("Можно получить паспорт")   # с отступом — внутри if

print("Это выполнится всегда")        # без отступа — уже снаружи
````

Поменяй `age = 15` на `age = 10` и запусти снова — первая строка не напечатается.

### Сравнения

| Запись | Значит |
|---|---|
| `a == b` | равно |
| `a != b` | не равно |
| `a > b`, `a < b` | больше, меньше |
| `a >= b`, `a <= b` | больше или равно, меньше или равно |
| `x in список` | есть в списке |

Не путай: `=` — **положить** значение в переменную, `==` — **сравнить**.

### `else` — «иначе»

````python
temperature = 8

if temperature >= 15:
    print("Можно идти в футболке")
else:
    print("Надень куртку")
````

Выполнится **одна** из двух веток: либо `if`, либо `else`.
У `else` условия нет — только двоеточие.

### `elif` — «а если нет, то может...»

Когда вариантов больше двух. Python проверяет сверху вниз и берёт **первую** подходящую ветку.

````python
mark = 4

if mark == 5:
    print("Отлично")
elif mark == 4:
    print("Хорошо")
elif mark == 3:
    print("Удовлетворительно")
else:
    print("Надо подтянуть")
````

### `if` вместе со списком

````python
guests = ["Аня", "Петя", "Маша"]

if "Петя" in guests:
    print("Петя приглашён")
else:
    print("Пети в списке нет")
````

### Частые ошибки с `if`

| Ошибка | Что напишет Python |
|---|---|
| забыл двоеточие: `if age > 10` | `SyntaxError: expected ':'` |
| забыл отступ внутри `if` | `IndentationError: expected an indented block` |
| написал `=` вместо `==`: `if age = 10:` | `SyntaxError` |

## 6. Цикл `for` — «для каждого»

Цикл берёт элементы списка **по одному** и для каждого выполняет код внутри.

```python
for переменная in список:
    что делать с каждым элементом
```

Те же правила: **двоеточие** в конце и **отступ** внутри.

````python
fruits = ["яблоко", "банан", "груша"]

for fruit in fruits:
    print("Я люблю", fruit)

print("Цикл закончился")   # без отступа — выполнится один раз, после цикла
````

### Как Python выполняет цикл по шагам

| Шаг | Чему равно `fruit` | Что напечатается |
|---|---|---|
| 1 | `"яблоко"` | Я люблю яблоко |
| 2 | `"банан"` | Я люблю банан |
| 3 | `"груша"` | Я люблю груша |
| — | элементы кончились | цикл завершён, идём дальше |

Имя переменной (`fruit`) придумываешь ты сам. Удобно: список во множественном числе
(`fruits`), переменная цикла — в единственном (`fruit`).

### `range` — цикл по числам

`range(n)` даёт числа от `0` до `n - 1`. **Как и в срезах, конец не входит.**

````python
for i in range(5):
    print(i)          # 0 1 2 3 4

print("---")
for i in range(1, 4):
    print(i)          # 1 2 3 — начинаем с 1, а 4 не входит
````

### Приём 1: накопление суммы

Заводим переменную **до** цикла (пока она `0`), а в цикле добавляем к ней по одному числу.
`total += price` — это короткая запись `total = total + price`.

````python
prices = [120, 45, 80]

total = 0
for price in prices:
    total += price
    print("Добавил", price, "→ сейчас", total)

print("Итого:", total)    # без отступа: печатаем один раз, когда цикл закончился
````

Разбор по шагам:

| Шаг | `price` | `total` было | `total` стало |
|---|---|---|---|
| 1 | 120 | 0 | 120 |
| 2 | 45 | 120 | 165 |
| 3 | 80 | 165 | 245 |

### Приём 2: счётчик

Сколько элементов подходят под условие? Счётчик стартует с `0` и увеличивается на `1`.
Здесь `for` и `if` работают вместе — **два уровня отступа**.

````python
marks = [5, 3, 4, 5, 2, 5]

count_fives = 0
for mark in marks:
    if mark == 5:            # отступ 1: внутри for
        count_fives += 1     # отступ 2: внутри if
print("Пятёрок:", count_fives)
````

### Приём 3: собрать новый список (фильтр)

Создаём **пустой** список `[]` и в цикле кладём в него подходящие элементы через `.append()`.

````python
numbers = [7, 12, 3, 20, 8, 15]

big = []
for number in numbers:
    if number > 10:
        big.append(number)
print(big)   # [12, 20, 15]
````

### Частые ошибки с `for`

1. **`print` результата с отступом.** Он напечатается на каждом шаге, а не один раз в конце.
2. **Счётчик внутри цикла.** Если `total = 0` поставить внутрь `for`, он будет обнуляться на каждом шаге.
3. **Забыл двоеточие или отступ** — как и с `if`.

## Шпаргалка

| Что нужно | Как написать |
|---|---|
| первые n элементов | `lst[:n]` |
| последние n элементов | `lst[-n:]` |
| добавить в конец / вставить | `lst.append(x)` / `lst.insert(i, x)` |
| удалить по значению | `lst.remove(x)` |
| есть ли элемент | `x in lst` |
| пройти по всем элементам | `for x in lst:` |
| числа от 0 до n-1 | `range(n)` |
| если / иначе если / иначе | `if ...:` / `elif ...:` / `else:` |

**Три правила:**
1. После `if`, `elif`, `else`, `for` — **двоеточие**, на следующей строке — **отступ**.
2. Счётчики и суммы заводим **до** цикла, печатаем результат **после** цикла.
3. В срезах и `range` **конец не входит**.
