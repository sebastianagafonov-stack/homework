## 1. Списочное включение

Включение создаёт коллекцию из правила преобразования и перебора. Запись `[expression for item in values if condition]` читается как «для подходящих item взять expression». Она удобна для простого отбора или преобразования.

Сначала полезно понимать обычный цикл с append, затем сокращать его до включения. Длинная вложенная логика или побочные действия внутри выражения ухудшают чтение; обычный цикл остаётся полноценным решением.


```python
values = [3, -1, 0, 4]
squares = [value * value for value in values if value > 0]
print(squares)
assert squares == [9, 16]
```

## 2. Включения для set и dict

Фигурные скобки с одним выражением создают множество; с записью ключ:значение — словарь. Так можно нормализовать уникальные слова или построить индекс данных.

Правила коллекций сохраняются: set теряет повторы, а одинаковый ключ dict оставляет последнее значение. Вложенные for выполняются в том же порядке, что и вложенные циклы. Сокращение синтаксиса не меняет смысл входных данных.


```python
words = [" API ", "sql", "api"]
unique = {word.strip().lower() for word in words}
lengths = {word: len(word) for word in ["api", "python"]}
flat = [value for row in [[1, 2], [3]] for value in row]
print(unique, lengths, flat)
```

**Практика:** Определить, почему после нормализации осталось два уникальных слова.


## 3. sorted и sort

sorted создаёт новый отсортированный список из последовательности или другой перебираемой коллекции. Метод list.sort меняет сам список и возвращает None. reverse=True задаёт убывание; это отличается от простого разворота несортированных данных.

Для понятного результата элементы должны быть сравнимы: смесь обычных чисел и строк без правила сравнения сортировать нельзя. Python выполняет стабильную сортировку: при равном ключе сохраняется исходный порядок элементов.


```python
values = [3, 1, 2]
ordered = sorted(values)
result = values.sort(reverse=True)
print(ordered, values, result)
words = ["aa", "b", "cc"]
print(sorted(words, key=len))
```

**Практика:** Сравнить sorted(values), values[::-1] и sort(reverse=True) на несортированном входе.


**Справочное примечание.** В key=len передаётся сама готовая операция len без вызова. Функции как значения и собственные ключи сортировки подробно разбираются в теме 26.


## 4. Сводные операции и пустые данные

sum, min и max вычисляют сумму, минимум и максимум. len даёт число элементов. Эти операции делают намерение понятнее, чем вручную написанный цикл, когда требуется только стандартный результат.

sum пустой последовательности равна 0. min и max без элементов вызывают ValueError; для перебираемой коллекции можно указать default, например `min(values, default=None)`. Среднее дополнительно требует проверки длины, чтобы не делить на ноль.


```python
values = []
print(sum(values), len(values), min(values, default=None))
values = [4, 2, 9]
print(sum(values), min(values), max(values))
```
