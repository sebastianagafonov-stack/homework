# Ссылки и копирование

**Результат:** различать равенство и идентичность и выбирать достаточный уровень копии.


## 1. Имя связано с объектом

Объект хранит значение, а имя ссылается на него. Присваивание другого имени тому же списку не создаёт новый список. Изменение объекта через одно имя видно через второе.

Это проявляется при подготовке версии данных, вложенных коллекциях и передаче объектов между частями программы. Присваивание имени нового объекта — другое действие: второе имя остаётся связано с прежним объектом.


```python
original = [1, 2]
alias = original
alias.append(3)
print(original, alias)
alias = [9]
print(original, alias)
assert original == [1, 2, 3]
```

**Практика:** Объяснить различие между alias.append(3) и alias = [9].


## 2. Равенство и идентичность

`==` проверяет равенство значений, `is` — тот же ли это объект. Два отдельных списка могут быть равны, но не идентичны. Сравнивать обычные строки и числа по значению следует через ==, а не через is.

Для None применяется `is None`, потому что проверяется специальный объект отсутствия значения. `id` возвращает идентификатор объекта на время его жизни; это средство наблюдения, не постоянный ключ для файла или базы данных.


```python
first = [1, 2]
second = [1, 2]
alias = first
print(first == second, first is second, first is alias)
print(id(first) == id(alias))
assert first == second and first is not second
```

**Практика:** Создать равные отдельные списки и подтвердить их независимость изменением одного.


## 3. Поверхностная копия

copy или срез списка создаёт новый внешний контейнер. В нём находятся ссылки на те же элементы. Для списка неизменяемых чисел независимого внешнего списка часто достаточно; для списка списков внутренние объекты остаются общими.

Такая копия называется поверхностной. Она полезна, если требуется независимо добавить или заменить внешнюю позицию, но не обещает независимость всех вложенных данных.


```python
source = [[1], [2]]
shallow = source.copy()
shallow.append([3])
shallow[0].append(9)
print(source, shallow)
assert source == [[1, 9], [2]]
assert source is not shallow and source[0] is shallow[0]
```

**Практика:** Различить добавление внешнего элемента и изменение первого вложенного списка.


## 4. Глубокая копия и выбор

deepcopy из стандартного модуля copy рекурсивно копирует вложенные объекты с учётом их структуры. Подключение `from copy import deepcopy` делает готовую операцию доступной по имени. Установка через pip не нужна: модуль входит в Python.

Глубокая копия нужна, когда изменяемые вложенные данные должны стать независимыми. Она требует дополнительных затрат и не является обязательным действием перед каждой операцией. Иногда проще создать новый результат из нужных полей, не копируя всё.


```python
from copy import deepcopy
source = {"marks": [4, 5]}
independent = deepcopy(source)
independent["marks"].append(3)
print(source, independent)
assert source == {"marks": [4, 5]}
```

**Практика:** Выбрать между присваиванием, copy и deepcopy для плоского списка и вложенной карточки.
