## 1. Вспоминаем списки

Список — это несколько значений в одной переменной, по порядку, в квадратных скобках.

## 2. Срезы — взять кусок списка

Срез пишется так: `список[начало:конец]`.
**Начало входит, конец — НЕ входит.** Получается новый список.

````python
days = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]

print(days[0:3])   # ['пн', 'вт', 'ср'] — индексы 0, 1, 2 (3 не входит!)
print(days[2:5])   # ['ср', 'чт', 'пт']
````

Начало или конец можно не писать:
- нет начала — берём **с самого начала**;
- нет конца — берём **до самого конца**.

````python
days = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]

print(days[:2])    # ['пн', 'вт'] — первые 2
print(days[4:])    # ['пт', 'сб', 'вс'] — начиная с индекса 4 и до конца
print(days[-2:])   # ['сб', 'вс'] — последние 2
print(days[:])     # весь список (это копия)
````

Удобное правило: **`список[:n]` — первые n элементов, `список[-n:]` — последние n.**

Срезы работают и со строками — точно так же:

````python
word = "программа"
print(word[:4])    # прог
print(word[-3:])   # мма
````

## 3. Как изменить список

Список **изменяемый**: элементы можно добавлять, удалять и заменять.

| Что сделать | Как | Пример |
|---|---|---|
| добавить в конец | `.append(x)` | `pets.append("хомяк")` |
| вставить на место | `.insert(индекс, x)` | `pets.insert(0, "рыбка")` |
| удалить по значению | `.remove(x)` | `pets.remove("кот")` |
| удалить по индексу и получить элемент | `.pop(индекс)` | `pets.pop(1)` |
| удалить последний | `.pop()` | `pets.pop()` |
| заменить | `список[индекс] = x` | `pets[0] = "попугай"` |

````python
pets = ["кот", "собака"]

pets.append("хомяк")       # в конец
print(pets)                # ['кот', 'собака', 'хомяк']

pets.insert(0, "рыбка")    # на место с индексом 0, остальные сдвигаются вправо
print(pets)                # ['рыбка', 'кот', 'собака', 'хомяк']

pets.remove("кот")         # удалить элемент "кот"
print(pets)                # ['рыбка', 'собака', 'хомяк']

last = pets.pop()          # вытащить последний — он сохранится в переменную
print(last)                # хомяк
print(pets)                # ['рыбка', 'собака']

pets[0] = "попугай"        # заменить
print(pets)                # ['попугай', 'собака']
````

**Важно:**
- `.remove(x)` удаляет только **первое** найденное `x`. Если `x` в списке нет — будет ошибка `ValueError`.
- Методы `.append`, `.insert`, `.remove` меняют список **на месте** и ничего не возвращают.
  Поэтому так писать **нельзя**: `pets = pets.append("хомяк")` — в `pets` окажется `None`.

## 4. Поиск в списке: `in`, `.count()`, `.index()`

````python
fruits = ["яблоко", "банан", "яблоко", "груша"]

print("банан" in fruits)        # True  — есть ли такой элемент?
print("киви" in fruits)         # False
print("киви" not in fruits)     # True  — «нет ли такого элемента?»

print(fruits.count("яблоко"))   # 2 — сколько раз встречается
print(fruits.index("груша"))    # 3 — на каком индексе (первое вхождение)
````

`in` даёт `True` или `False`, поэтому его очень удобно использовать в условиях `if` — об этом дальше.

## 5. Условие `if` — «если»

Программа не всегда должна делать одно и то же. Иногда нужно:
**«если условие верно — сделай это»**.

```python
if условие:
    код, который выполнится, только если условие True
```

Три обязательные вещи:
1. слово `if`;
2. **двоеточие** `:` в конце строки;
3. **отступ** (4 пробела или Tab) у кода внутри. Именно отступ показывает Python,
   какие строки относятся к `if`.

````python
age = 15

if age >= 14:
    print("Можно получить паспорт")   # с отступом — внутри if
    print("Это тоже внутри if")

print("А это выполнится всегда")      # без отступа — уже не внутри if
````

Поменяй в ячейке выше `age = 15` на `age = 10` и запусти снова — первые две строки не напечатаются.

### Сравнения (повторение)

| Запись | Значит |
|---|---|
| `a == b` | равно |
| `a != b` | не равно |
| `a > b`, `a < b` | больше, меньше |
| `a >= b`, `a <= b` | больше или равно, меньше или равно |
| `x in список` | есть в списке |

Не путай: `=` — **положить** значение в переменную, `==` — **сравнить**.

### `else` — «иначе»

Если нужно сделать что-то, когда условие **не** выполнилось:

````python
temperature = 8

if temperature >= 15:
    print("Можно идти в футболке")
else:
    print("Надень куртку")
````

Выполнится **ровно одна** из двух веток: либо та, что под `if`, либо та, что под `else`.
У `else` нет условия — только двоеточие.

### `elif` — «а если нет, то может...»

Когда вариантов больше двух:

````python
mark = 4

if mark == 5:
    print("Отлично")
elif mark == 4:
    print("Хорошо")
elif mark == 3:
    print("Удовлетворительно")
else:
    print("Надо подтянуть")
````

Как это работает: Python проверяет условия **сверху вниз** и выполняет **первую** ветку,
у которой условие `True`. Остальные уже не проверяются.

Поэтому порядок важен. Пример с ошибкой:

````python
score = 95

# НЕПРАВИЛЬНО: 95 >= 50 — это True, поэтому до проверки >= 90 дело не дойдёт
if score >= 50:
    print("Зачёт")
elif score >= 90:
    print("Отлично!")

# ПРАВИЛЬНО: сначала самое строгое условие
if score >= 90:
    print("Отлично!")
elif score >= 50:
    print("Зачёт")
````

### Несколько условий сразу: `and`, `or`, `not`

- `and` — «и»: нужно, чтобы **оба** условия были верны;
- `or` — «или»: достаточно **одного**;
- `not` — «не»: переворачивает `True` ↔ `False`.

````python
age = 15
has_ticket = True

if age >= 12 and has_ticket:
    print("Проходи в кино")

day = "сб"
if day == "сб" or day == "вс":
    print("Выходной!")

if not has_ticket:
    print("Сначала купи билет")   # не напечатается: has_ticket = True, а not True = False
````

## 6. Цикл `for` — «для каждого»

Цикл повторяет код несколько раз. Цикл `for` берёт элементы списка **по одному**
и для каждого выполняет код внутри.

```python
for переменная in список:
    код, который выполнится для каждого элемента
```

Те же правила, что у `if`: двоеточие в конце и отступ у кода внутри.

````python
fruits = ["яблоко", "банан", "груша"]

for fruit in fruits:
    print("Я люблю", fruit)

print("Цикл закончился")   # без отступа — выполнится один раз, после цикла
````

Что происходит по шагам:
1. `fruit = "яблоко"` → печатаем «Я люблю яблоко»;
2. `fruit = "банан"` → печатаем «Я люблю банан»;
3. `fruit = "груша"` → печатаем «Я люблю груша»;
4. элементы кончились — цикл завершён, идём дальше.

Имя переменной (`fruit`) придумываешь ты сам. Хорошая привычка: список во множественном
числе (`fruits`), переменная цикла — в единственном (`fruit`).

### `range` — цикл по числам

`range(n)` даёт числа от `0` до `n - 1` — **как в срезах, конец не входит**.

````python
for i in range(5):
    print(i)          # 0 1 2 3 4 — каждое на новой строке

print("---")
for i in range(1, 4):
    print(i)          # 1 2 3 — от 1 до 4, но 4 не входит
````

Посмотреть, какие числа даёт `range`, можно так: `list(range(1, 4))` → `[1, 2, 3]`.

### Пример 1: посчитать сумму (накопление)

Заводим переменную **до** цикла, а в цикле понемногу добавляем к ней.

````python
prices = [120, 45, 80]

total = 0                 # до цикла: пока ничего не набрали
for price in prices:
    total = total + price # можно короче: total += price
    print("Добавил", price, "→ сейчас", total)

print("Итого:", total)    # после цикла (без отступа!) печатаем результат
````

Конечно, для суммы есть `sum(prices)`. Но этот Пример работает для чего угодно:
посчитать, склеить, собрать — и без него не обойтись в задачах посложнее.

### Пример 2: счётчик

Сколько элементов подходят под условие? Счётчик начинается с `0` и увеличивается на `1`.
Здесь `for` и `if` работают **вместе** — обрати внимание на **два уровня отступа**.

````python
marks = [5, 3, 4, 5, 2, 5]

count_fives = 0
for mark in marks:
    if mark == 5:            # отступ 1: внутри for
        count_fives += 1     # отступ 2: внутри if
print("Пятёрок:", count_fives)
````

### Пример 3: собрать новый список (фильтр)

Создаём **пустой** список `[]` и в цикле добавляем в него подходящие элементы через `.append()`.

````python
numbers = [7, 12, 3, 20, 8, 15]

big = []
for number in numbers:
    if number > 10:
        big.append(number)
print(big)   # [12, 20, 15]
````

### Пример 4: изменить каждый элемент

````python
prices = [100, 250, 40]

with_discount = []
for price in prices:
    with_discount.append(price // 2)   # скидка 50%
print(with_discount)   # [50, 125, 20]
````

### Пример 5: нужен номер элемента

Если нужен и номер, и сам элемент, используй `enumerate`. Через `start=1` нумерация начнётся с единицы:

````python
tasks = ["сделать ДЗ", "погулять с собакой", "убрать комнату"]

for number, task in enumerate(tasks, start=1):
    print(f"{number}. {task}")
````

### Частые ошибки с `for`

1. **`print` результата внутри цикла.** Если написать `print("Итого:", total)` с отступом,
   он напечатается на **каждом** шаге, а не один раз в конце. Отступ решает всё.
2. **Счётчик внутри цикла.** Если `total = 0` поставить внутрь `for`, он будет обнуляться на каждом шаге.
3. **Удаление элементов из списка, по которому идёт цикл.** Так делать нельзя — элементы «перескакиваются».
   Вместо этого собери новый список (Пример 3).

````python
# Ошибка 1 — print внутри цикла: «Итого» напечатается 3 раза
prices = [120, 45, 80]
total = 0
for price in prices:
    total += price
    print("Итого:", total)
````

## Шпаргалка

| Что нужно | Как написать |
|---|---|
| первые n элементов | `lst[:n]` |
| последние n элементов | `lst[-n:]` |
| кусок с индекса a до b (b не входит) | `lst[a:b]` |
| добавить в конец / вставить | `lst.append(x)` / `lst.insert(i, x)` |
| удалить по значению / по индексу | `lst.remove(x)` / `lst.pop(i)` |
| есть ли элемент | `x in lst` |
| сколько раз встречается | `lst.count(x)` |
| пройти по всем элементам | `for x in lst:` |
| числа от 0 до n-1 | `range(n)` |
| если / иначе если / иначе | `if ...:` / `elif ...:` / `else:` |

**Три правила:**
1. После `if`, `elif`, `else`, `for` — **двоеточие**, на следующей строке — **отступ**.
2. Счётчики и суммы заводим **до** цикла, печатаем результат **после** цикла.
3. В срезах и `range` **конец не входит**.
